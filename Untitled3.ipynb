```python
import pandas as pd
import glob

# Find the uploaded CSV files
customers_file = glob.glob('/content/customers-100000*.csv')[0]
people_file = glob.glob('/content/people-100000*.csv')[0]
organizations_file = glob.glob('/content/organizations-100000*.csv')[0]
leads_file = glob.glob('/content/leads-100000*.csv')[0]
products_file = glob.glob('/content/products-100000*.csv')[0]

# Load the datasets
customers = pd.read_csv(customers_file)
people = pd.read_csv(people_file)
organizations = pd.read_csv(organizations_file)
leads = pd.read_csv(leads_file)
products = pd.read_csv(products_file)

# Check row counts and column names
datasets = {
    "Customers": customers,
    "People": people,
    "Organizations": organizations,
    "Leads": leads,
    "Products": products
}

for name, data in datasets.items():
    print("\nDataset:", name)
    print("Number of rows:", len(data))
    print("Number of columns:", len(data.columns))
    print("Column names:", data.columns.tolist())
```

```python
import pandas as pd

# Create cleaned copies of the five datasets
cleaned_datasets = {}
cleaning_summary = []

for name, df in datasets.items():

    # Make a copy so the original data stays unchanged
    clean_df = df.copy()

    # Remove extra spaces from column names
    clean_df.columns = [str(col).strip() for col in clean_df.columns]

    # Remove extra spaces from text values
    for col in clean_df.select_dtypes(include=["object", "string"]).columns:
        clean_df[col] = clean_df[col].map(
            lambda value: value.strip()
            if isinstance(value, str) else value
        )

        # Treat empty text cells as missing values
        clean_df[col] = clean_df[col].replace(r"^\s*$", pd.NA, regex=True)

    # Keep the cleaned dataset for the next steps
    cleaned_datasets[name] = clean_df

    # Count missing values and exact duplicate rows
    missing_count = clean_df.isna().sum().sum()
    duplicate_count = clean_df.duplicated().sum()

    cleaning_summary.append({
        "Dataset": name,
        "Rows": clean_df.shape[0],
        "Columns": clean_df.shape[1],
        "Missing Cells": missing_count,
        "Exact Duplicate Rows": duplicate_count
    })

# Display the data quality summary
summary_df = pd.DataFrame(cleaning_summary)

print("DATA CLEANING SUMMARY")
display(summary_df)
```

```python
for name, df in cleaned_datasets.items():
    missing = df.isna().sum()
    missing = missing[missing > 0].sort_values(ascending=False)

    print("\nDataset:", name)

    if missing.empty:
        print("No missing values found.")
    else:
        print(missing)
```

```python
# Step 4: Check repeated values in important columns

key_columns = {
    "Customers": ["Customer Id", "Email", "Website"],
    "People": ["User Id", "Email", "Phone"],
    "Organizations": ["Organization Id", "Website", "Name"],
    "Leads": ["Account Id", "Email", "Website"],
    "Products": ["EAN", "Internal ID", "Name"]
}

duplicate_results = []

for dataset_name, columns in key_columns.items():
    df = cleaned_datasets[dataset_name]

    for column in columns:
        if column in df.columns:
            values = df[column].dropna().astype(str).str.strip()
            counts = values.value_counts()
            repeated = counts[counts > 1]

            duplicate_results.append({
                "Dataset": dataset_name,
                "Column": column,
                "Unique Values": values.nunique(),
                "Repeated Values": len(repeated),
                "Rows in Repeated Groups": int(repeated.sum()),
                "Extra Repeated Rows": int((repeated - 1).sum())
            })

duplicate_df = pd.DataFrame(duplicate_results)

print("DUPLICATE CHECK RESULTS")
display(duplicate_df)
```

```python
# STEP 5: Common columns and common data

from itertools import combinations

# Show common column names for every pair of datasets
print("COMMON COLUMNS")

for (name1, df1), (name2, df2) in combinations(
    cleaned_datasets.items(), 2
):
    common_columns = sorted(set(df1.columns) & set(df2.columns))
    print(f"\n{name1} + {name2}: {common_columns}")


# Standardize text before comparing values
def clean_text(column):
    return (
        column.fillna("")
        .astype(str)
        .str.strip()
        .str.lower()
    )


# Standardize website text
def clean_website(column):
    website = clean_text(column)
    website = website.str.replace(
        r"^https?://", "", regex=True
    )
    website = website.str.replace(
        r"^www\.", "", regex=True
    )
    website = website.str.rstrip("/")
    website = website.str.split("/").str[0]
    return website


# Count common values and possible join combinations
def compare_columns(left, right, description):
    left = left[left != ""]
    right = right[right != ""]

    common_values = set(left) & set(right)

    left_counts = left.value_counts()
    right_counts = right.value_counts()

    left_rows = int(left.isin(common_values).sum())
    right_rows = int(right.isin(common_values).sum())

    joined_pairs = sum(
        int(left_counts[value]) * int(right_counts[value])
        for value in common_values
    )

    print("\n", description)
    print("Distinct common values:", len(common_values))
    print("Rows matched in first dataset:", left_rows)
    print("Rows matched in second dataset:", right_rows)
    print("Possible joined row pairs:", joined_pairs)


# Email comparisons
compare_columns(
    clean_text(cleaned_datasets["Customers"]["Email"]),
    clean_text(cleaned_datasets["People"]["Email"]),
    "Customers vs People — Email"
)

compare_columns(
    clean_text(cleaned_datasets["Customers"]["Email"]),
    clean_text(cleaned_datasets["Leads"]["Email"]),
    "Customers vs Leads — Email"
)

compare_columns(
    clean_text(cleaned_datasets["People"]["Email"]),
    clean_text(cleaned_datasets["Leads"]["Email"]),
    "People vs Leads — Email"
)


# Website comparisons
compare_columns(
    clean_website(cleaned_datasets["Customers"]["Website"]),
    clean_website(cleaned_datasets["Organizations"]["Website"]),
    "Customers vs Organizations — Website"
)

compare_columns(
    clean_website(cleaned_datasets["Leads"]["Website"]),
    clean_website(cleaned_datasets["Organizations"]["Website"]),
    "Leads vs Organizations — Website"
)

compare_columns(
    clean_website(cleaned_datasets["Customers"]["Website"]),
    clean_website(cleaned_datasets["Leads"]["Website"]),
    "Customers vs Leads — Website"
)


# Company name comparisons
compare_columns(
    clean_text(cleaned_datasets["Customers"]["Company"]),
    clean_text(cleaned_datasets["Leads"]["Company"]),
    "Customers vs Leads — Company"
)

compare_columns(
    clean_text(cleaned_datasets["Customers"]["Company"]),
    clean_text(cleaned_datasets["Organizations"]["Name"]),
    "Customers vs Organizations — Company/Name"
)

compare_columns(
    clean_text(cleaned_datasets["Leads"]["Company"]),
    clean_text(cleaned_datasets["Organizations"]["Name"]),
    "Leads vs Organizations — Company/Name"
)
```

```python
# STEP 7: FINAL ANALYSIS REPORT

from itertools import combinations
from IPython.display import display
import pandas as pd

# 1. Data quality summary
quality_report = []

for name, df in cleaned_datasets.items():
    quality_report.append({
        "Dataset": name,
        "Rows": len(df),
        "Columns": len(df.columns),
        "Missing Cells": int(df.isna().sum().sum()),
        "Exact Duplicate Rows": int(df.duplicated().sum())
    })

quality_report = pd.DataFrame(quality_report)

print("1. DATA QUALITY SUMMARY")
display(quality_report)


# 2. Repeated values in important columns
key_columns = {
    "Customers": ["Customer Id", "Email", "Website"],
    "People": ["User Id", "Email", "Phone"],
    "Organizations": ["Organization Id", "Website"],
    "Leads": ["Account Id", "Email", "Website"],
    "Products": ["EAN", "Internal ID"]
}

duplicate_report = []

for name, columns in key_columns.items():
    df = cleaned_datasets[name]

    for column in columns:
        if column in df.columns:
            values = df[column].dropna().astype(str).str.strip()
            counts = values.value_counts()
            repeated = counts[counts > 1]

            duplicate_report.append({
                "Dataset": name,
                "Column": column,
                "Unique Values": values.nunique(),
                "Repeated Values": len(repeated),
                "Extra Repeated Rows": int((repeated - 1).sum())
            })

duplicate_report = pd.DataFrame(duplicate_report)

print("\n2. DUPLICATE VALUE ANALYSIS")
display(duplicate_report)


# 3. Common columns between datasets
common_columns_report = []

for (name1, df1), (name2, df2) in combinations(
    cleaned_datasets.items(), 2
):
    common = sorted(set(df1.columns) & set(df2.columns))

    common_columns_report.append({
        "Dataset 1": name1,
        "Dataset 2": name2,
        "Common Columns": ", ".join(common)
    })

common_columns_report = pd.DataFrame(common_columns_report)

print("\n3. COMMON COLUMNS")
display(common_columns_report)


# 4. Check matching values across selected datasets
def normalize_text(series):
    return series.fillna("").astype(str).str.strip().str.lower()


def normalize_website(series):
    website = normalize_text(series)
    website = website.str.replace(r"^https?://", "", regex=True)
    website = website.str.replace(r"^www\.", "", regex=True)
    website = website.str.rstrip("/")
    return website.str.split("/").str[0]


def check_matches(df1, col1, df2, col2, match_type):
    left = normalize_website(df1[col1]) if match_type == "website" \
        else normalize_text(df1[col1])
    right = normalize_website(df2[col2]) if match_type == "website" \
        else normalize_text(df2[col2])

    left = left[left != ""]
    right = right[right != ""]

    common = set(left.unique()) & set(right.unique())
    left_counts = left.value_counts()
    right_counts = right.value_counts()

    joined_pairs = sum(
        int(left_counts[value]) * int(right_counts[value])
        for value in common
    )

    return {
        "Distinct Matching Values": len(common),
        "Rows Matched in Dataset 1": int(left.isin(common).sum()),
        "Rows Matched in Dataset 2": int(right.isin(common).sum()),
        "Possible Joined Row Pairs": joined_pairs
    }


c = cleaned_datasets["Customers"]
p = cleaned_datasets["People"]
o = cleaned_datasets["Organizations"]
l = cleaned_datasets["Leads"]

join_checks = []

comparisons = [
    ("Customers", c, "Email", "People", p, "Email", "text"),
    ("Customers", c, "Email", "Leads", l, "Email", "text"),
    ("People", p, "Email", "Leads", l, "Email", "text"),
    ("Customers", c, "Website", "Organizations", o, "Website", "website"),
    ("Leads", l, "Website", "Organizations", o, "Website", "website"),
    ("Customers", c, "Website", "Leads", l, "Website", "website"),
    ("Customers", c, "Company", "Leads", l, "Company", "text"),
    ("Customers", c, "Company", "Organizations", o, "Name", "text"),
    ("Leads", l, "Company", "Organizations", o, "Name", "text")
]

for name1, df1, col1, name2, df2, col2, kind in comparisons:
    result = check_matches(df1, col1, df2, col2, kind)

    join_checks.append({
        "Comparison": f"{name1}.{col1} vs {name2}.{col2}",
        **result
    })

join_checks = pd.DataFrame(join_checks)

print("\n4. COMMON DATA AND JOIN CHECKS")
display(join_checks)


# 5. Business questions and calculated answers
answers = []

def add_answer(question, answer, observation):
    answers.append({
        "Business Question": question,
        "Answer": answer,
        "Observation": observation
    })


# Customers
countries = c["Country"].value_counts().head(5)
add_answer(
    "Which countries have the most customers?",
    str(countries.to_dict()),
    "Congo and Korea have the highest customer counts."
)

subscription_dates = pd.to_datetime(c["Subscription Date"], errors="coerce")
years = subscription_dates.dt.year.value_counts().sort_index()

add_answer(
    "How many customers subscribed each year?",
    str(years.to_dict()),
    "2022 covers only part of the year in this dataset."
)

# People
sex_counts = p["Sex"].value_counts()
add_answer(
    "What is the gender distribution?",
    str(sex_counts.to_dict()),
    "The counts for female and male records are close."
)

add_answer(
    "Which job titles are most common?",
    str(p["Job Title"].value_counts().head(5).to_dict()),
    "No single job title dominates the records."
)

# Organizations
add_answer(
    "Which industries appear most often?",
    str(o["Industry"].value_counts().head(5).to_dict()),
    "Insurance is the most common industry."
)

employees = pd.to_numeric(o["Number of employees"], errors="coerce")
add_answer(
    "What is the typical organization size?",
    f"Average: {employees.mean():.2f}; Median: {employees.median():.0f}; "
    f"Minimum: {employees.min():.0f}; Maximum: {employees.max():.0f}",
    "Organization sizes vary considerably."
)

# Leads
add_answer(
    "Which companies have the most leads?",
    str(l["Company"].value_counts().head(5).to_dict()),
    "Repeated company names can represent multiple leads for one company."
)

# Products
add_answer(
    "How many products are in each availability status?",
    str(products["Availability"].value_counts().to_dict()),
    "The six availability categories have broadly similar counts."
)

price = pd.to_numeric(products["Price"], errors="coerce")
add_answer(
    "What is the product price range?",
    f"Average: {price.mean():.2f}; Median: {price.median():.0f}; "
    f"Minimum: {price.min():.0f}; Maximum: {price.max():.0f}",
    "These are product list prices, not actual sales revenue."
)

stock = pd.to_numeric(products["Stock"], errors="coerce")
add_answer(
    "What is the total product stock?",
    f"Total: {stock.sum():.0f}; Average: {stock.mean():.2f}; "
    f"Minimum: {stock.min():.0f}; Maximum: {stock.max():.0f}",
    "Stock quantities vary across the product records."
)

add_answer(
    "Which product categories have the most products?",
    str(products["Category"].value_counts().head(5).to_dict()),
    "Women's Clothing and Books & Stationery are tied at the top."
)

out_of_stock_with_stock = (
    (normalize_text(products["Availability"]) == "out_of_stock")
    & (stock > 0)
).sum()

add_answer(
    "Are any out-of-stock products showing positive stock?",
    str(int(out_of_stock_with_stock)),
    "These records should be checked for inconsistent stock status."
)

add_answer(
    "Is Internal ID unique in Products?",
    f"{products['Internal ID'].nunique()} unique values across "
    f"{len(products)} rows",
    "Internal ID should not be treated as a unique product key."
)

business_answers = pd.DataFrame(answers)

print("\n5. BUSINESS QUESTIONS AND ANSWERS")
display(business_answers)


# 6. Final observations
email_customers_leads = check_matches(
    c, "Email", l, "Email", "text"
)["Distinct Matching Values"]

final_observations = pd.DataFrame({
    "Final Observations": [
        "All five datasets contain 100,000 rows.",
        "Initial cleaning found no missing cells or exact duplicate rows.",
        "Repeated IDs and contact values must be checked separately from exact duplicate rows.",
        f"Customers and Leads share {email_customers_leads} email values; "
        "the matching records should be checked for identity conflicts.",
        "Website and company-name matching can produce many-to-many joins.",
        "Do not join datasets on Index because it is only a row number.",
        "Products contains 100,000 distinct EAN values but only 99 distinct Internal ID values.",
        f"{int(out_of_stock_with_stock)} products marked out_of_stock have positive stock.",
        "These are synthetic practice datasets, not verified real-business performance data."
    ]
})

print("\n6. FINAL ANALYSIS")
display(final_observations)


# 7. Save all results in one Excel workbook
output_file = "/content/Final_Data_Analysis_Report.xlsx"

with pd.ExcelWriter(output_file) as writer:
    quality_report.to_excel(writer, sheet_name="Data Quality", index=False)
    duplicate_report.to_excel(writer, sheet_name="Duplicates", index=False)
    common_columns_report.to_excel(writer, sheet_name="Common Columns", index=False)
    join_checks.to_excel(writer, sheet_name="Join Checks", index=False)
    business_answers.to_excel(writer, sheet_name="Business Answers", index=False)
    final_observations.to_excel(writer, sheet_name="Final Analysis", index=False)

print("\nReport saved to:", output_file)
```
